```python
# Create a Python program that helps a school IT team monitor login attempts.

# Your program should:

# 1. Ask for the analyst’s name.
# 2. Repeatedly show a menu until the user exits.
# 3. Allow the user to:

#    * Add a login attempt
#    * View all login attempts
#    * Count failed logins by username
#    * Count failed logins by IP address
#    * Flag suspicious IPs
#    * Exit
# 4. Store login attempts using a list of dictionaries.
# 5. Use functions to organize the program.
# 6. Use input validation.
# 7. Use loops, conditionals, lists, dictionaries, and string handling.

# 1. input analyst's name
# 2. loop with exit condition
# 3. could be different functions - add a login attempt,  view all login attempts, ..
# 4. functions - 
# add login attempt - use dictionary {username: login status, ip address} if they don't add all 3, give an error
# view - print
# count by username - what if the user or IP address has no failed attempts + message
# count by ip address
# flag suspicious IPs - any IP that has 2 or more failed logins is suspicious
# if analysts exits - goodbye message + analyst name

#wrapper function
def main_func():

    #initialize the list
    login_attempts = []

    #taking analyst's name and validating it
    print("Please enter analyst's name:")
    analyst = input()
    if analyst.strip() == "":
        print("Analyst name cannot be blank")
        return
        
    #running through the menu and calling functions
    while True:
        print("Please enter your desired option:\n Press 1 for adding a login attempt\n Press 2 for viewing all login attempts\n Press 3 to count failed login attempts by username\n Press 4 to count failed login attempts by IP address\n Press 5 to view suspicious IP addresses (any IPs with over 2 failed login attempts)\n Press 6 to exit")
        option = input()

        if option == '1':
            add_login(login_attempts)
        elif option == '2':
            view_login(login_attempts)
        elif option == '3':
            count_username(login_attempts)
        elif option == '4':
            count_ip(login_attempts)
        elif option == '5':
            sus_login(login_attempts)
        elif option == '6':
            print("Thank you! See you again, ", analyst)
            break
        #if analyst enters invalid option
        else:
            print("Please enter a valid option")
    return

#function to add login attempts
def add_login(login_attempts):

    #add username
    print("Please enter the username")
    username = input()
    if username == "":
        print("Please enter a valid username")
        return

    #add login status
    print("Please enter the login status. Note: Login status can only be success or failure")
    status = input().lower().strip()
    if status != "failure" and status.lower().strip() != "success":
        print("Please enter the correct status")
        return

    #add IP Address
    print("Please enter the IP Address")
    ip = input()
    if ip.strip() == "":
        print("Please enter a valid IP Address")
        return

    #adding all the inputs in a dictionary and then adding them to the main list
    d = {'username': username, 'status': status, 'ip': ip}
    login_attempts.append(d)
    print("Login attempt was successfully recorded")

def view_login(login_attempts):

    #iterating through the list first and then printing key, value of the dictionary
    if len(login_attempts) != 0:
        for i in login_attempts:
            for key,value in i.items():
                print(f"{key} : {value}")
    else:
        print("No login attempts have been recorded")

def count_username(login_attempts):

    #initialize the dictionary needed for counting
    count_username = {}

    #checking for failed login attempts and adding them to the dictionary above
    for i in login_attempts:
        if i['status'] == "failure":
            username = i['username']

            #if this is the first time we are seeing this username, we need to "initialize" it in the dictionary
            if username not in count_username:
                count_username[username] = 0
                
            #and then increase the counter each time we see the failure status
            count_username[username] += 1

    if len(count_username) != 0:
        for key,value in count_username.items():
            print(f" User {key} has {value} number of failed login attempts")
    else:
        print("No failed login attempts have been recorded")


def count_ip(login_attempts):

    #initialize the dictionary needed for counting
    count_ip = {}

    #checking for failed login attempts and adding them to the dictionary above
    for i in login_attempts:
        if i['status'] == "failure":
            ip = i['ip']

            #if this is the first time we are seeing this IP, we need to "initialize" it in the dictionary
            if ip not in count_ip:
                count_ip[ip] = 0
                
            #and then increase the counter each time we see the failure status
            count_ip[ip] += 1

    if len(count_ip) != 0:
        for key,value in count_ip.items():
            print(f" IP address {key} has {value} number of failed login attempts")
    else:
        print("No failed login attempts have been recorded")
    

def sus_login(login_attempts):

    #initializing sus IP list
    sus_ip = []
    
   #initialize the dictionary needed for counting
    count_ip = {}

    #checking for failed login attempts and adding them to the dictionary above
    for i in login_attempts:
        if i['status'] == "failure":
            ip = i['ip']

            #if this is the first time we are seeing this IP, we need to "initialize" it in the dictionary
            if ip not in count_ip:
                count_ip[ip] = 0
                
            #and then increase the counter each time we see the failure status
            count_ip[ip] += 1

    for i in count_ip:
        if count_ip['ip'] >= 2:
            sus_ip.append(ip)

    #printing sus IPs
    if len(sus_ip) != 0:
        print(sus_ip)
    else:
        print("No suspicious IPs were identified")
```
